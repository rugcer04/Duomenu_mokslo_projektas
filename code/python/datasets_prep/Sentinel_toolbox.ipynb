```python
import os
import pandas as pd
from pathlib import Path
```

```python
base_path = Path(os.getcwd())
repo_root = base_path.parents[3]
print(f"Current working directory: {base_path}")
print(f"Repository root directory: {repo_root}")
```

```python
with open(os.path.join(repo_root, "sentinel_data", "lithuania_aerosol_index_vilnius.csv"), 'r') as f:
    df_aerosol = pd.read_csv(f)

with open(os.path.join(repo_root, "sentinel_data", "lithuania_cloud_vilnius.csv"), 'r') as f:
    df_cloud = pd.read_csv(f)

with open(os.path.join(repo_root, "sentinel_data", "lithuania_nitrogen_vilnius.csv"), 'r') as f:
    df_nitrogen = pd.read_csv(f)

with open(os.path.join(repo_root, "sentinel_data", "lithuania_ozone_vilnius.csv"), 'r') as f:
    df_ozone = pd.read_csv(f)

# with open(os.path.join(repo_root, "sentinel_data", "lithuania_sulfur_vilnius.csv"), 'r') as f:
#     df_sulfur = pd.read_csv(f)

# with open(os.path.join(repo_root, "sentinel_data", "lithuania_methane_vilnius.csv"), 'r') as f:
#     df_methane = pd.read_csv(f)

# with open(os.path.join(repo_root, "sentinel_data", "lithuania_formaldehyde_vilnius.csv"), 'r') as f:
#     df_formaldehyde = pd.read_csv(f)

with open(os.path.join(repo_root, "sentinel_data", "lithuania_carbon_vilnius.csv"), 'r') as f:
    df_carbon = pd.read_csv(f)
```

```python
# print(df_aerosol.head(10))
# print(df_cloud.head(10))
# print(df_nitrogen.head(10))
# print(df_ozone.head(10))
```

```python
dfs = [df_aerosol, df_cloud, df_nitrogen, df_ozone, df_carbon]

cleaned_dfs = []

for df in dfs:
    # df['time'] = pd.to_datetime(df['time'])
    
    if df['time'].duplicated().any():
        print(f"Duplicate timestamps found! Keeping the first occurrence.")
        df = df.drop_duplicates(subset='time')
    
    df = df.set_index('time')
    cleaned_dfs.append(df)
```

```python
df_combined = pd.concat(cleaned_dfs, axis=1)
df_combined = df_combined.reset_index()
```

```python
# Check for missing hours in the sequence
expected_range = pd.date_range(start=df_combined.index.min(), 
                               end=df_combined.index.max(), 
                               freq='h')

missing_count = len(expected_range) - len(df_combined.index)
print(f"Total missing hourly timestamps: {missing_count}")
```

```python
df_combined.to_csv("Sentinel_5p_data.csv")
```
