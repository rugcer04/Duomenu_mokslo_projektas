```python
from pathlib import Path
import xarray as xr
import os
import dask
```

```python
base_path = Path(os.getcwd())
repo_root = base_path.parents[2]

data_dir = Path(f"{repo_root}/DataGRIB")
files = sorted(list(data_dir.glob("*.grib")))

ds = xr.open_mfdataset(
    files, 
    engine='cfgrib', 
    combine='nested', 
    concat_dim='time',
    backend_kwargs={'indexpath': ''}
)
```
