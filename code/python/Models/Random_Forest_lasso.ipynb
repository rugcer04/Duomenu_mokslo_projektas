```python
def create_lags(column_name, lag_value, df_train, df_val, df_test):

    new_col_name = f"{column_name}_lag-{lag_value}"
    
    df_train.loc[:, new_col_name] = df_train[column_name].shift(lag_value)
    df_val.loc[:, new_col_name] = df_val[column_name].shift(lag_value)
    df_test.loc[:, new_col_name] = df_test[column_name].shift(lag_value)
```

```python
X_train.columns
```

```python
lags_info = {
    'absorbing_aerosol_index': [8, 19],
    'NO2_column_number_density': [1, 2, 3],
    'stratospheric_NO2_column_number_density': [1, 3, 7],
    'NO2_slant_column_number_density': [1, 2],
    'tropopause_pressure': [1],
    'H2O_column_number_density': [1, 2],
    'NO2': [1, 4, 9],
    'dust': [1, 4],
    'sia': [11],
    'pm2p5': [1, 11],
    'pans': [10, 18],
    'nmvoc': [1, 2],
    'nh3': [1, 11],
    'pm10': [1, 11]
}


for col, lags in lags_info.items():
    if isinstance(lags, list):
        for lag in lags:
            create_lags(col, lag, train, val, test)
    else:
        create_lags(col, lags, train, val, test)
```

```python
def rolling_window_fun(column_name, window_value, df_train, df_val, df_test):

    new_col_name = f"{column_name}_window-{window_value}"
    
    df_train.loc[:, new_col_name] = df_train[column_name].rolling(window=window_value).mean()
    df_val.loc[:, new_col_name] = df_val[column_name].rolling(window=window_value).mean()
    df_test.loc[:, new_col_name] = df_test[column_name].rolling(window=window_value).mean()

window_info = {
    # 'absorbing_aerosol_index': [8, 19],
    'NO2_column_number_density': [3, 2],
    'stratospheric_NO2_column_number_density': [3, 5, 7],
    'NO2_slant_column_number_density': [2],
    # 'tropopause_pressure': [1],
    'H2O_column_number_density': [2],
    'NO2': [3, 5, 7],
    'dust': [4, 6, 8],
    'sia': [11],
    # 'pm2p5': [1, 11],
    'pans': [10],
    'nmvoc': [3, 5],
    'nh3': [2, 4],
    'pm10': [5]
}


for col, windows in window_info.items():
    if isinstance(windows, list):
        for window in windows:
            rolling_window_fun(col, window, train, val, test)
    else:
        rolling_window_fun(col, windows, train, val, test)
```

### Realizacija

```python
rf_default = RandomForestRegressor(random_state=123)
rf_default.fit(X_train, y_train)
```

```python
rf_default_pred_val = rf_default.predict(X_val)
```

### Metrikos

```python
def metrics_fun(y_val, rf_pred_val):
    print(f'R^2:  {r2_score(y_val, rf_pred_val)}')
    print(f'MAE:  {mean_absolute_error(y_val, rf_pred_val)}')
    print(f'RMSE: {np.sqrt(mean_squared_error(y_val, rf_pred_val))}')
    # print(f'MAPE: {mean_absolute_percentage_error(y_val, rf_pred_val)}')

    corr, p_value = stats.spearmanr(y_val, rf_pred_val)
    print(f"r_s:  {corr:.3f}")
```

```python
metrics_fun(y_val, rf_default_pred_val)
```

### Požymių reikšmingumas

```python
rf_default_importances = rf_default.feature_importances_
rf_default_importances
```

```python
result = permutation_importance(rf_default, X_test, y_test, n_repeats=5, random_state=123)
sorted_importances_idx = result.importances_mean.argsort()
importances_df = pd.DataFrame(
    result.importances[sorted_importances_idx].T,
    columns=X_test.columns[sorted_importances_idx]
)

forest_importances = pd.Series(result.importances_mean, index=X_test.columns)
forest_importances = forest_importances[forest_importances > 0]
forest_importances = forest_importances.sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(12, 6))
forest_importances.plot.barh(ax=ax, color='skyblue')
ax.set_title("Permutation Importance (mean)")
ax.set_xlabel("Mean decrease in accuracy score")
ax.axvline(x=0, color="k", linestyle="--")

fig.tight_layout()
plt.show()
```
